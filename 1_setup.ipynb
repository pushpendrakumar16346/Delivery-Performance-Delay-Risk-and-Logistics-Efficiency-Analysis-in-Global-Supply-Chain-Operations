```python
from pathlib import Path
import pandas as pd
import numpy as np
import json
import shutil
import sys

PROJECT_ROOT = Path(r"C:\Users\hp\Analysis in Global Supply Chain Operations")

DIRECTORIES = [
    "data/raw",
    "data/interim",
    "data/processed",
    "notebooks",
    "src/data_processing",
    "src/analysis",
    "src/visualization",
    "src/utilities",
    "dashboard/pages",
    "outputs/figures",
    "outputs/tables",
    "outputs/reports",
    "research_paper",
    "executive_summary",
    "documentation",
    "tests",
    "final_submission"
]

for directory in DIRECTORIES:
    (PROJECT_ROOT / directory).mkdir(parents=True, exist_ok=True)

print("=" * 75)
print("PROJECT DIRECTORY SETUP")
print("=" * 75)

print(f"\nProject Name: {PROJECT_ROOT.name}")
print(f"Project Location: {PROJECT_ROOT}")

uploaded_file = Path("APL_Logistics.xlsx")
raw_file = PROJECT_ROOT / "data" / "raw" / "APL_Logistics.xlsx"

if uploaded_file.exists():
    shutil.copy2(uploaded_file, raw_file)
elif raw_file.exists():
    print("\nDataset already exists in the project directory.")
else:
    possible_locations = [
        Path.cwd() / "data" / "raw" / "APL_Logistics.xlsx",
        Path.cwd().parent / "APL_Logistics.xlsx",
        Path(r"C:\Users\hp\APL_Logistics.xlsx")
    ]

    for location in possible_locations:
        if location.exists():
            shutil.copy2(location, raw_file)
            break

if not raw_file.exists():
    raise FileNotFoundError(
        "APL_Logistics.xlsx was not found. Place the uploaded dataset "
        "in the project data/raw folder and run this cell again."
    )

print("\nDATASET INFORMATION")
print("-" * 75)

df = pd.read_excel(raw_file, engine="openpyxl")

expected_columns = [
    "Type",
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Benefit per order",
    "Sales per customer",
    "Delivery Status",
    "Late_delivery_risk",
    "Category Id",
    "Category Name",
    "Customer City",
    "Customer Country",
    "Customer Fname",
    "Customer Id",
    "Customer Lname",
    "Customer Segment",
    "Customer State",
    "Customer Street",
    "Customer Zipcode",
    "Department Id",
    "Department Name",
    "Latitude",
    "Longitude",
    "Market",
    "Order City",
    "Order Country",
    "Order Customer Id",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Order Item Quantity",
    "Sales",
    "Order Item Total",
    "Order Profit Per Order",
    "Order Region",
    "Order State",
    "Order Status",
    "Product Name",
    "Product Price",
    "Shipping Mode"
]

missing_columns = [
    column for column in expected_columns
    if column not in df.columns
]

print(f"Dataset Path: {raw_file}")
print(f"Dataset Shape: {df.shape}")
print(f"Number of Records: {len(df):,}")
print(f"Number of Columns: {len(df.columns)}")
print(f"Expected Columns: {len(expected_columns)}")
print(f"Missing Expected Columns: {missing_columns}")

if missing_columns:
    raise ValueError(
        "Dataset validation failed because expected columns are missing."
    )

print("\nCOLUMN DATA TYPES")
display(df.dtypes.to_frame("Data Type"))

print("\nFIRST FIVE RECORDS")
display(df.head())

print("\nDATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
missing_values = df.isnull().sum()
display(
    missing_values[missing_values > 0]
    .to_frame("Missing Values")
)

print("\nNUMERICAL SUMMARY")
display(df.describe().T)

print("\nCATEGORICAL SUMMARY")
display(df.describe(include="object").T)

date_columns = [
    column for column in df.columns
    if any(
        keyword in column.lower()
        for keyword in ["date", "datetime", "timestamp"]
    )
]

print("\nDATE COLUMN AVAILABILITY")
print(date_columns if date_columns else "No date column found.")

configuration = {
    "project_name": "Analysis in Global Supply Chain Operations",
    "project_root": str(PROJECT_ROOT),
    "dataset_path": str(raw_file),
    "dataset_name": "APL_Logistics.xlsx",
    "target_column": "Late_delivery_risk",
    "shipping_actual_column": "Days for shipping (real)",
    "shipping_scheduled_column": "Days for shipment (scheduled)",
    "expected_columns": expected_columns,
    "random_seed": 42
}

with open(
    PROJECT_ROOT / "config.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(configuration, file, indent=4)

requirements = [
    "pandas",
    "numpy",
    "openpyxl",
    "matplotlib",
    "seaborn",
    "plotly",
    "streamlit",
    "scikit-learn",
    "statsmodels",
    "scipy",
    "jupyter"
]

with open(
    PROJECT_ROOT / "requirements.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write("\n".join(requirements))

print("\nPROJECT DIRECTORY CONTENTS")
for item in sorted(PROJECT_ROOT.iterdir()):
    print(item.name)

print("\n" + "=" * 75)
print("PROJECT SETUP COMPLETED")
print("=" * 75)

print(f"Project Location: {PROJECT_ROOT}")
print(f"Dataset Shape: {df.shape}")
print("Dataset Column Validation: PASS")
print("Configuration File: CREATED")
print("Requirements File: CREATED")
print("Project Directories: CREATED")
```
