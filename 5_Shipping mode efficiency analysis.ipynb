```python
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

warnings.filterwarnings("ignore")

print("=" * 90)
print("SHIPPING MODE EFFICIENCY ANALYSIS")
print("=" * 90)


# LOAD AND INSPECT DELIVERY GAP DATASET

possible_roots = [
    Path.cwd(),
    *Path.cwd().parents,
    Path(r"C:\Users\hp\Analysis in Global Supply Chain Operations")
]

PROJECT_ROOT = None

for candidate in possible_roots:
    expected_file = candidate / "data" / "processed" / "delivery_gap_analysis.csv"
    if expected_file.exists():
        PROJECT_ROOT = candidate
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Could not locate data\\processed\\delivery_gap_analysis.csv. "
        "Check the project directory and dataset location."
    )

INPUT_FILE = PROJECT_ROOT / "data" / "processed" / "delivery_gap_analysis.csv"

OUTPUT_DIRS = {
    "tables": PROJECT_ROOT / "outputs" / "tables",
    "reports": PROJECT_ROOT / "outputs" / "reports",
    "figures": PROJECT_ROOT / "outputs" / "figures"
}

for directory in OUTPUT_DIRS.values():
    directory.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(INPUT_FILE)

required_columns = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Delivery Status",
    "Late_delivery_risk",
    "Shipping Mode"
]

missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    raise ValueError(f"Required columns are missing: {missing_columns}")

print(f"\nProject directory: {PROJECT_ROOT}")
print(f"Input dataset: {INPUT_FILE}")
print(f"Dataset rows: {len(df):,}")
print(f"Dataset columns: {len(df.columns):,}")

print("\nAvailable shipping modes:")
print(df["Shipping Mode"].value_counts(dropna=False).to_string())


# VALIDATE SHIPPING MODE AND DELIVERY DURATION DATA

analysis_df = df.copy()

analysis_df["Shipping Mode"] = (
    analysis_df["Shipping Mode"]
    .astype("string")
    .str.strip()
)

analysis_df["Days for shipping (real)"] = pd.to_numeric(
    analysis_df["Days for shipping (real)"],
    errors="coerce"
)

analysis_df["Days for shipment (scheduled)"] = pd.to_numeric(
    analysis_df["Days for shipment (scheduled)"],
    errors="coerce"
)

analysis_df["Late_delivery_risk"] = pd.to_numeric(
    analysis_df["Late_delivery_risk"],
    errors="coerce"
)

analysis_df["Delivery Status"] = (
    analysis_df["Delivery Status"]
    .astype("string")
    .str.strip()
)

analysis_df["Shipping Mode"] = analysis_df["Shipping Mode"].replace(
    {"": pd.NA, "nan": pd.NA, "None": pd.NA}
)

analysis_df["Delivery Status"] = analysis_df["Delivery Status"].replace(
    {"": pd.NA, "nan": pd.NA, "None": pd.NA}
)

analysis_df["Delay Gap"] = (
    analysis_df["Days for shipping (real)"]
    - analysis_df["Days for shipment (scheduled)"]
)

analysis_df["Is Cancelled"] = (
    analysis_df["Delivery Status"]
    .str.casefold()
    .eq("shipping canceled")
    .fillna(False)
)

analysis_df["Valid Shipping Duration"] = (
    analysis_df["Days for shipping (real)"].notna()
    & analysis_df["Days for shipment (scheduled)"].notna()
    & (analysis_df["Days for shipping (real)"] >= 0)
    & (analysis_df["Days for shipment (scheduled)"] >= 0)
)

analysis_df["Valid Shipping Mode"] = analysis_df["Shipping Mode"].notna()

analysis_df["Eligible Delivery"] = (
    ~analysis_df["Is Cancelled"]
    & analysis_df["Valid Shipping Duration"]
    & analysis_df["Valid Shipping Mode"]
)

analysis_df["Delivery Classification"] = np.select(
    [
        analysis_df["Delay Gap"] > 0,
        analysis_df["Delay Gap"] == 0,
        analysis_df["Delay Gap"] < 0
    ],
    [
        "Delayed",
        "On-time",
        "Early"
    ],
    default="Invalid"
)

analysis_df.loc[
    ~analysis_df["Eligible Delivery"],
    "Delivery Classification"
] = "Ineligible"

eligible_df = analysis_df.loc[
    analysis_df["Eligible Delivery"]
].copy()

if eligible_df.empty:
    raise ValueError("No eligible delivery records are available for analysis.")

shipping_modes = sorted(
    eligible_df["Shipping Mode"].dropna().unique().tolist()
)

print("\nDATA VALIDATION SUMMARY")
print("-" * 90)
print(f"Total records: {len(analysis_df):,}")
print(f"Eligible delivery records: {len(eligible_df):,}")
print(f"Cancelled shipments: {int(analysis_df['Is Cancelled'].sum()):,}")
print(
    "Invalid or incomplete duration records:",
    int((~analysis_df["Valid Shipping Duration"]).sum())
)
print(
    "Records without a valid shipping mode:",
    int((~analysis_df["Valid Shipping Mode"]).sum())
)
print(f"Identified shipping modes: {len(shipping_modes)}")


# CALCULATE MODE-WISE SHIPMENT VOLUMES AND DELIVERY PERFORMANCE

comparison_records = []

for mode in shipping_modes:

    all_mode_records = analysis_df[
        analysis_df["Shipping Mode"] == mode
    ].copy()

    mode_df = eligible_df[
        eligible_df["Shipping Mode"] == mode
    ].copy()

    total_shipments = len(all_mode_records)
    eligible_shipments = len(mode_df)

    delayed_count = int(
        (mode_df["Delivery Classification"] == "Delayed").sum()
    )

    on_time_count = int(
        (mode_df["Delivery Classification"] == "On-time").sum()
    )

    early_count = int(
        (mode_df["Delivery Classification"] == "Early").sum()
    )

    cancelled_count = int(all_mode_records["Is Cancelled"].sum())

    delayed_percentage = (
        delayed_count / eligible_shipments * 100
        if eligible_shipments else np.nan
    )

    on_time_percentage = (
        on_time_count / eligible_shipments * 100
        if eligible_shipments else np.nan
    )

    early_percentage = (
        early_count / eligible_shipments * 100
        if eligible_shipments else np.nan
    )

    risk_valid = mode_df["Late_delivery_risk"].isin([0, 1])

    risk_ratio = (
        mode_df.loc[risk_valid, "Late_delivery_risk"].mean() * 100
        if risk_valid.any() else np.nan
    )

    actual_mean = mode_df["Days for shipping (real)"].mean()
    scheduled_mean = mode_df["Days for shipment (scheduled)"].mean()
    average_gap = mode_df["Delay Gap"].mean()

    median_gap = mode_df["Delay Gap"].median()

    average_delay_days = (
        mode_df.loc[
            mode_df["Delivery Classification"] == "Delayed",
            "Delay Gap"
        ].mean()
    )

    maximum_delay_days = (
        mode_df.loc[
            mode_df["Delivery Classification"] == "Delayed",
            "Delay Gap"
        ].max()
    )

    # SLA compliance means actual shipping duration did not exceed scheduled duration.
    sla_compliant_count = int(
        (mode_df["Delay Gap"] <= 0).sum()
    )

    sla_compliance_percentage = (
        sla_compliant_count / eligible_shipments * 100
        if eligible_shipments else np.nan
    )

    comparison_records.append({
        "Shipping Mode": mode,
        "Total Shipments": total_shipments,
        "Eligible Shipments": eligible_shipments,
        "Cancelled Shipments": cancelled_count,
        "Delayed Shipments": delayed_count,
        "On-time Shipments": on_time_count,
        "Early Shipments": early_count,
        "Average Actual Shipping Days": actual_mean,
        "Average Scheduled Shipping Days": scheduled_mean,
        "Average Delay Gap": average_gap,
        "Median Delay Gap": median_gap,
        "Average Delay Among Delayed Shipments": average_delay_days,
        "Maximum Delay Gap": maximum_delay_days,
        "Delayed Percentage": delayed_percentage,
        "On-time Percentage": on_time_percentage,
        "Early Delivery Percentage": early_percentage,
        "Late Delivery Risk Ratio": risk_ratio,
        "SLA Compliant Shipments": sla_compliant_count,
        "SLA Compliance Percentage": sla_compliance_percentage,
        "Risk Records Available": int(risk_valid.sum())
    })

mode_comparison = pd.DataFrame(comparison_records)

print("\nMODE-WISE SHIPPING PERFORMANCE")
print("-" * 90)
display(mode_comparison.round(3))


# GENERATE MODE-WISE SLA COMPLIANCE REPORT

sla_report = mode_comparison[
    [
        "Shipping Mode",
        "Total Shipments",
        "Eligible Shipments",
        "SLA Compliant Shipments",
        "SLA Compliance Percentage",
        "Delayed Shipments",
        "Delayed Percentage",
        "On-time Percentage",
        "Early Delivery Percentage"
    ]
].copy()

sla_report["SLA Status"] = np.select(
    [
        sla_report["SLA Compliance Percentage"] >= 95,
        sla_report["SLA Compliance Percentage"] >= 85
    ],
    [
        "High SLA Compliance",
        "Moderate SLA Compliance"
    ],
    default="Low SLA Compliance"
)

print("\nMODE-WISE SLA COMPLIANCE REPORT")
print("-" * 90)
display(sla_report.round(3))


# COMPARE INDIVIDUAL SHIPPING MODES

print("\nINDIVIDUAL SHIPPING MODE COMPARISON")
print("-" * 90)

expected_modes = [
    "First Class",
    "Second Class",
    "Same Day",
    "Standard Class"
]

for mode in expected_modes:

    mode_result = mode_comparison[
        mode_comparison["Shipping Mode"].str.casefold() == mode.casefold()
    ]

    if mode_result.empty:
        print(f"\n{mode}: No records available.")
        continue

    result = mode_result.iloc[0]

    print(f"\n{mode}")
    print(f"Total shipments: {int(result['Total Shipments']):,}")
    print(f"Eligible shipments: {int(result['Eligible Shipments']):,}")
    print(f"Delayed shipments: {int(result['Delayed Shipments']):,}")
    print(f"Average actual shipping days: {result['Average Actual Shipping Days']:.3f}")
    print(f"Average scheduled shipping days: {result['Average Scheduled Shipping Days']:.3f}")
    print(f"Average delay gap: {result['Average Delay Gap']:.3f}")
    print(f"Delayed percentage: {result['Delayed Percentage']:.2f}%")
    print(f"On-time percentage: {result['On-time Percentage']:.2f}%")
    print(f"Early delivery percentage: {result['Early Delivery Percentage']:.2f}%")
    print(f"Late delivery risk ratio: {result['Late Delivery Risk Ratio']:.2f}%")
    print(f"SLA compliance: {result['SLA Compliance Percentage']:.2f}%")


# ANALYSE DELIVERY STATUS DISTRIBUTION BY SHIPPING MODE

status_comparison = pd.crosstab(
    analysis_df["Shipping Mode"],
    analysis_df["Delivery Status"],
    dropna=False
)

status_comparison = status_comparison.reindex(shipping_modes)

status_percentage = status_comparison.div(
    status_comparison.sum(axis=1).replace(0, np.nan),
    axis=0
) * 100

status_percentage = status_percentage.add_suffix(" (%)")

print("\nDELIVERY STATUS COUNTS BY SHIPPING MODE")
print("-" * 90)
display(status_comparison)

print("\nDELIVERY STATUS PERCENTAGES BY SHIPPING MODE")
print("-" * 90)
display(status_percentage.round(2))


# ANALYSE DELAY GAP DISTRIBUTION BY SHIPPING MODE

delay_gap_summary = (
    eligible_df.groupby("Shipping Mode")["Delay Gap"]
    .agg(
        Shipment_Count="count",
        Mean_Delay_Gap="mean",
        Median_Delay_Gap="median",
        Minimum_Delay_Gap="min",
        Maximum_Delay_Gap="max",
        Standard_Deviation="std"
    )
    .reset_index()
)

delay_gap_summary = delay_gap_summary.sort_values(
    "Mean_Delay_Gap",
    ascending=False
)

print("\nDELAY GAP COMPARISON BY SHIPPING MODE")
print("-" * 90)
display(delay_gap_summary.round(3))


# IDENTIFY HIGHEST DELAY FREQUENCY AND BEST-PERFORMING MODE

highest_delay_frequency = mode_comparison.sort_values(
    ["Delayed Percentage", "Delayed Shipments"],
    ascending=[False, False]
).iloc[0]

best_sla_mode = mode_comparison.sort_values(
    ["SLA Compliance Percentage", "Late Delivery Risk Ratio"],
    ascending=[False, True]
).iloc[0]

best_on_time_mode = mode_comparison.sort_values(
    "On-time Percentage",
    ascending=False
).iloc[0]

highest_risk_mode = mode_comparison.sort_values(
    "Late Delivery Risk Ratio",
    ascending=False
).iloc[0]

print("\nSHIPPING MODE PERFORMANCE FINDINGS")
print("-" * 90)

print(
    f"Highest delay frequency: {highest_delay_frequency['Shipping Mode']} "
    f"({highest_delay_frequency['Delayed Percentage']:.2f}%)"
)

print(
    f"Best SLA-compliant mode: {best_sla_mode['Shipping Mode']} "
    f"({best_sla_mode['SLA Compliance Percentage']:.2f}%)"
)

print(
    f"Highest on-time percentage: {best_on_time_mode['Shipping Mode']} "
    f"({best_on_time_mode['On-time Percentage']:.2f}%)"
)

print(
    f"Highest late delivery risk: {highest_risk_mode['Shipping Mode']} "
    f"({highest_risk_mode['Late Delivery Risk Ratio']:.2f}%)"
)


# CALCULATE SHIPPING MODE EFFICIENCY INDEX

# The index combines SLA compliance, the existing late-risk indicator,
# and relative average delay-gap performance.
# All percentage components use eligible shipments as their denominator.
# The delay-gap component is scaled across the observed shipping modes.

gap_min = mode_comparison["Average Delay Gap"].min()
gap_max = mode_comparison["Average Delay Gap"].max()

if pd.isna(gap_min) or pd.isna(gap_max):
    mode_comparison["Delay Gap Efficiency Score"] = np.nan
elif np.isclose(gap_min, gap_max):
    mode_comparison["Delay Gap Efficiency Score"] = 100.0
else:
    mode_comparison["Delay Gap Efficiency Score"] = (
        (gap_max - mode_comparison["Average Delay Gap"])
        / (gap_max - gap_min)
    ) * 100

mode_comparison["SLA Efficiency Score"] = (
    mode_comparison["SLA Compliance Percentage"]
)

mode_comparison["Risk Efficiency Score"] = (
    100 - mode_comparison["Late Delivery Risk Ratio"]
)

mode_comparison["Shipping Mode Efficiency Index"] = (
    0.50 * mode_comparison["SLA Efficiency Score"]
    + 0.30 * mode_comparison["Risk Efficiency Score"]
    + 0.20 * mode_comparison["Delay Gap Efficiency Score"]
)

mode_comparison["Efficiency Rank"] = (
    mode_comparison["Shipping Mode Efficiency Index"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

mode_comparison["Efficiency Category"] = np.select(
    [
        mode_comparison["Shipping Mode Efficiency Index"] >= 80,
        mode_comparison["Shipping Mode Efficiency Index"] >= 60
    ],
    [
        "High Efficiency",
        "Moderate Efficiency"
    ],
    default="Low Efficiency"
)

efficiency_ranking = mode_comparison[
    [
        "Efficiency Rank",
        "Shipping Mode",
        "Shipping Mode Efficiency Index",
        "SLA Compliance Percentage",
        "Late Delivery Risk Ratio",
        "Average Delay Gap",
        "Delay Gap Efficiency Score",
        "Efficiency Category"
    ]
].sort_values(
    ["Efficiency Rank", "Shipping Mode"]
).reset_index(drop=True)

print("\nSHIPPING MODE EFFICIENCY INDEX")
print("-" * 90)
print(
    "Index formula: 50% SLA compliance + 30% inverse late-risk ratio "
    "+ 20% relative delay-gap score."
)
print(
    "Note: The delay-gap score is scaled across the shipping modes in this dataset; "
    "it is a relative comparison, not an absolute service standard."
)
display(efficiency_ranking.round(3))


# IDENTIFY INEFFICIENT SHIPPING MODES

inefficient_modes = efficiency_ranking[
    efficiency_ranking["Efficiency Category"] == "Low Efficiency"
].copy()

if inefficient_modes.empty:
    print("\nNo shipping modes fall into the low-efficiency category.")
else:
    print("\nSHIPPING MODES REQUIRING PERFORMANCE IMPROVEMENT")
    print("-" * 90)
    display(inefficient_modes.round(3))


# GENERATE COMPARATIVE VISUALIZATIONS

sns.set_theme(style="whitegrid")

plot_df = mode_comparison.sort_values(
    "Shipping Mode"
).copy()

plt.figure(figsize=(11, 6))
sns.barplot(
    data=plot_df,
    x="Shipping Mode",
    y="Delayed Percentage"
)
plt.title("Delayed Shipment Percentage by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Delayed Shipments (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIRS["figures"] / "shipping_mode_delayed_percentage.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()


plt.figure(figsize=(11, 6))
sns.barplot(
    data=plot_df,
    x="Shipping Mode",
    y="On-time Percentage"
)
plt.title("On-time Delivery Percentage by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("On-time Shipments (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIRS["figures"] / "shipping_mode_on_time_percentage.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()


plt.figure(figsize=(11, 6))
sns.barplot(
    data=plot_df,
    x="Shipping Mode",
    y="SLA Compliance Percentage"
)
plt.title("SLA Compliance by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("SLA Compliance (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIRS["figures"] / "shipping_mode_sla_compliance.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()


plt.figure(figsize=(11, 6))
sns.barplot(
    data=plot_df,
    x="Shipping Mode",
    y="Average Delay Gap"
)
plt.axhline(0, linestyle="--", linewidth=1)
plt.title("Average Delivery Delay Gap by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Average Delay Gap (Days)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIRS["figures"] / "shipping_mode_average_delay_gap.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()


plt.figure(figsize=(11, 6))
sns.barplot(
    data=plot_df,
    x="Shipping Mode",
    y="Late Delivery Risk Ratio"
)
plt.title("Late Delivery Risk Ratio by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Late Delivery Risk (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIRS["figures"] / "shipping_mode_late_delivery_risk.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()


plt.figure(figsize=(11, 6))
sns.barplot(
    data=efficiency_ranking,
    x="Shipping Mode",
    y="Shipping Mode Efficiency Index"
)
plt.title("Shipping Mode Efficiency Index Comparison")
plt.xlabel("Shipping Mode")
plt.ylabel("Efficiency Index (0–100)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig(
    OUTPUT_DIRS["figures"] / "shipping_mode_efficiency_index.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()


# GENERATE DELIVERY STATUS COMPARISON VISUALIZATION

status_plot = status_comparison.copy()

status_plot.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 7)
)

plt.title("Delivery Status Distribution by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Shipment Count")
plt.xticks(rotation=20)
plt.legend(title="Delivery Status", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.savefig(
    OUTPUT_DIRS["figures"] / "shipping_mode_delivery_status_comparison.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
plt.close()


# GENERATE SHIPPING MODE EFFICIENCY REPORT

report_rows = []

for _, row in efficiency_ranking.iterrows():

    report_rows.append({
        "Shipping Mode": row["Shipping Mode"],
        "Efficiency Rank": int(row["Efficiency Rank"]),
        "Efficiency Index": float(row["Shipping Mode Efficiency Index"]),
        "Efficiency Category": row["Efficiency Category"],
        "SLA Compliance Percentage": float(row["SLA Compliance Percentage"]),
        "Late Delivery Risk Ratio": float(row["Late Delivery Risk Ratio"]),
        "Average Delay Gap": float(row["Average Delay Gap"]),
        "Interpretation": (
            "Strong relative performance"
            if row["Efficiency Rank"] == 1
            else (
                "Requires operational review"
                if row["Efficiency Category"] == "Low Efficiency"
                else "Moderate relative performance"
            )
        )
    })

efficiency_report = pd.DataFrame(report_rows)

print("\nSHIPPING MODE EFFICIENCY REPORT")
print("-" * 90)
display(efficiency_report.round(3))


# SAVE SHIPPING MODE ANALYSIS OUTPUTS

output_files = {}

output_files["comparison"] = (
    OUTPUT_DIRS["tables"] / "shipping_mode_comparison.csv"
)

output_files["sla"] = (
    OUTPUT_DIRS["tables"] / "shipping_mode_sla_report.csv"
)

output_files["ranking"] = (
    OUTPUT_DIRS["tables"] / "shipping_mode_efficiency_ranking.csv"
)

output_files["status_counts"] = (
    OUTPUT_DIRS["tables"] / "shipping_mode_delivery_status_counts.csv"
)

output_files["status_percentages"] = (
    OUTPUT_DIRS["tables"] / "shipping_mode_delivery_status_percentages.csv"
)

output_files["delay_gap"] = (
    OUTPUT_DIRS["tables"] / "shipping_mode_delay_gap_summary.csv"
)

output_files["efficiency_report"] = (
    OUTPUT_DIRS["reports"] / "shipping_mode_efficiency_report.csv"
)

output_files["individual_mode"] = (
    OUTPUT_DIRS["reports"] / "shipping_mode_individual_comparison.csv"
)

output_files["eligible_records"] = (
    OUTPUT_DIRS["reports"] / "shipping_mode_eligible_delivery_records.csv"
)

output_files["validation"] = (
    OUTPUT_DIRS["reports"] / "shipping_mode_analysis_validation.csv"
)

mode_comparison.to_csv(output_files["comparison"], index=False)
sla_report.to_csv(output_files["sla"], index=False)
efficiency_ranking.to_csv(output_files["ranking"], index=False)
status_comparison.to_csv(output_files["status_counts"])
status_percentage.to_csv(output_files["status_percentages"])
delay_gap_summary.to_csv(output_files["delay_gap"], index=False)
efficiency_report.to_csv(output_files["efficiency_report"], index=False)
mode_comparison.to_csv(output_files["individual_mode"], index=False)

eligible_df.to_csv(
    output_files["eligible_records"],
    index=False
)

# Define all generated chart filenames for validation

figure_files = [
    "shipping_mode_delayed_percentage.png",
    "shipping_mode_on_time_percentage.png",
    "shipping_mode_sla_compliance.png",
    "shipping_mode_average_delay_gap.png",
    "shipping_mode_late_delivery_risk.png",
    "shipping_mode_efficiency_index.png",
    "shipping_mode_delivery_status_comparison.png"
]
# VALIDATE SHIPPING MODE ANALYSIS

# VALIDATE SHIPPING MODE ANALYSIS

validation_results = []

def add_validation(check_name, condition, details):
    validation_results.append({
        "Validation Check": check_name,
        "Status": "PASS" if bool(condition) else "FAIL",
        "Details": str(details)
    })


add_validation(
    "Input dataset loaded",
    len(analysis_df) == len(df) and len(df) > 0,
    f"Loaded {len(df):,} records."
)

add_validation(
    "All shipping modes identified",
    len(shipping_modes) > 0
    and mode_comparison["Shipping Mode"].nunique() == len(shipping_modes),
    f"{len(shipping_modes)} shipping modes identified."
)

add_validation(
    "Shipment counts reconcile",
    int(mode_comparison["Total Shipments"].sum())
    == int(analysis_df["Valid Shipping Mode"].sum()),
    "Shipping mode shipment totals reconcile."
)

add_validation(
    "Eligible shipment counts reconcile",
    int(mode_comparison["Eligible Shipments"].sum()) == len(eligible_df),
    "Eligible shipment counts reconcile."
)

add_validation(
    "Delivery classifications reconcile",
    (
        int(mode_comparison["Delayed Shipments"].sum())
        + int(mode_comparison["On-time Shipments"].sum())
        + int(mode_comparison["Early Shipments"].sum())
    ) == len(eligible_df),
    "All eligible shipments are classified."
)

for _, row in mode_comparison.iterrows():

    denominator = row["Eligible Shipments"]

    if denominator > 0:
        add_validation(
            f"Percentage denominator: {row['Shipping Mode']}",
            np.isclose(
                row["Delayed Percentage"],
                row["Delayed Shipments"] / denominator * 100
            )
            and np.isclose(
                row["On-time Percentage"],
                row["On-time Shipments"] / denominator * 100
            )
            and np.isclose(
                row["Early Delivery Percentage"],
                row["Early Shipments"] / denominator * 100
            ),
            "All delivery percentages use eligible shipments."
        )

        add_validation(
            f"SLA reconciliation: {row['Shipping Mode']}",
            np.isclose(
                row["SLA Compliance Percentage"],
                row["SLA Compliant Shipments"] / denominator * 100
            ),
            "SLA compliance uses eligible shipments."
        )

add_validation(
    "Cancelled shipments excluded from eligible metrics",
    not eligible_df["Is Cancelled"].any(),
    "Cancelled shipments are excluded from delivery-performance denominators."
)

add_validation(
    "Efficiency index within valid range",
    mode_comparison["Shipping Mode Efficiency Index"].between(0, 100).all(),
    "Efficiency scores are within 0 and 100."
)

add_validation(
    "Efficiency ranking complete",
    mode_comparison["Efficiency Rank"].notna().all(),
    "All shipping modes have an efficiency ranking."
)

add_validation(
    "Expected shipping modes available",
    all(mode in shipping_modes for mode in expected_modes),
    "All four expected shipping modes are available."
)

# Verify all outputs except the validation report itself.

for name, file_path in output_files.items():

    if name == "validation":
        continue

    add_validation(
        f"Output file exists: {file_path.name}",
        file_path.exists() and file_path.stat().st_size > 0,
        f"Verified: {file_path}"
    )

for figure_name in figure_files:

    figure_path = OUTPUT_DIRS["figures"] / figure_name

    add_validation(
        f"Chart exists: {figure_name}",
        figure_path.exists() and figure_path.stat().st_size > 0,
        f"Verified: {figure_path}"
    )

# Save the validation report before checking its existence.

validation_report = pd.DataFrame(validation_results)

validation_report.to_csv(
    output_files["validation"],
    index=False
)

add_validation(
    "Output file exists: shipping_mode_analysis_validation.csv",
    output_files["validation"].exists()
    and output_files["validation"].stat().st_size > 0,
    f"Verified: {output_files['validation']}"
)

# Save the final validation report including its own file verification.

validation_report = pd.DataFrame(validation_results)

validation_report.to_csv(
    output_files["validation"],
    index=False
)

passed_checks = int(
    (validation_report["Status"] == "PASS").sum()
)

failed_checks = int(
    (validation_report["Status"] == "FAIL").sum()
)

print("\nSHIPPING MODE ANALYSIS VALIDATION")
print("-" * 90)
display(validation_report)

print(f"\nValidation checks passed: {passed_checks}")
print(f"Validation checks failed: {failed_checks}")


# SAVE ANALYSIS SUMMARY

summary = {
    "project": "Analysis in Global Supply Chain Operations",
    "analysis": "Shipping Mode Efficiency Analysis",
    "total_records": int(len(analysis_df)),
    "shipping_modes": shipping_modes,
    "eligible_shipments": int(len(eligible_df)),
    "cancelled_shipments": int(analysis_df["Is Cancelled"].sum()),
    "delayed_shipments": int(mode_comparison["Delayed Shipments"].sum()),
    "on_time_shipments": int(mode_comparison["On-time Shipments"].sum()),
    "early_shipments": int(mode_comparison["Early Shipments"].sum()),
    "highest_delay_frequency_mode": str(
        highest_delay_frequency["Shipping Mode"]
    ),
    "best_sla_compliance_mode": str(best_sla_mode["Shipping Mode"]),
    "highest_late_delivery_risk_mode": str(
        highest_risk_mode["Shipping Mode"]
    ),
    "best_efficiency_mode": str(
        efficiency_ranking.iloc[0]["Shipping Mode"]
    ),
    "validation_passed": passed_checks,
    "validation_failed": failed_checks,
    "validation_status": "PASS" if failed_checks == 0 else "FAIL"
}

summary_path = (
    OUTPUT_DIRS["reports"] / "shipping_mode_efficiency_summary.json"
)

with open(summary_path, "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=4, ensure_ascii=False)

print("\n" + "=" * 90)

if failed_checks == 0:
    print("SHIPPING MODE EFFICIENCY ANALYSIS COMPLETED SUCCESSFULLY")
    print("ALL VALIDATION CHECKS PASSED")
else:
    print("SHIPPING MODE EFFICIENCY ANALYSIS HAS VALIDATION FAILURES")

print("=" * 90)

# SAVE ANALYSIS SUMMARY

summary = {
    "project": "Analysis in Global Supply Chain Operations",
    "analysis": "Shipping Mode Efficiency Analysis",
    "source_file": str(INPUT_FILE),
    "total_records": int(len(analysis_df)),
    "shipping_modes": shipping_modes,
    "total_eligible_shipments": int(len(eligible_df)),
    "total_cancelled_shipments": int(analysis_df["Is Cancelled"].sum()),
    "total_delayed_shipments": int(
        mode_comparison["Delayed Shipments"].sum()
    ),
    "total_on_time_shipments": int(
        mode_comparison["On-time Shipments"].sum()
    ),
    "total_early_shipments": int(
        mode_comparison["Early Shipments"].sum()
    ),
    "highest_delay_frequency_mode": str(
        highest_delay_frequency["Shipping Mode"]
    ),
    "best_sla_compliance_mode": str(
        best_sla_mode["Shipping Mode"]
    ),
    "highest_late_delivery_risk_mode": str(
        highest_risk_mode["Shipping Mode"]
    ),
    "best_efficiency_mode": str(
        efficiency_ranking.iloc[0]["Shipping Mode"]
    ),
    "efficiency_index_formula": (
        "50% SLA compliance + 30% inverse late delivery risk "
        "+ 20% relative average delay-gap score"
    ),
    "validation_passed": passed_checks,
    "validation_failed": failed_checks,
    "validation_status": (
        "PASS" if failed_checks == 0 else "FAIL"
    ),
    "output_files": {
        key: str(path) for key, path in output_files.items()
    },
    "figure_files": [
        str(OUTPUT_DIRS["figures"] / name)
        for name in figure_files
    ]
}

summary_path = (
    OUTPUT_DIRS["reports"] / "shipping_mode_efficiency_summary.json"
)

with open(summary_path, "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=4, ensure_ascii=False)


# DISPLAY FINAL RESULTS

print("\nFINAL SHIPPING MODE EFFICIENCY RANKING")
print("=" * 90)
display(
    efficiency_ranking[
        [
            "Efficiency Rank",
            "Shipping Mode",
            "Shipping Mode Efficiency Index",
            "SLA Compliance Percentage",
            "Late Delivery Risk Ratio",
            "Efficiency Category"
        ]
    ].round(3)
)

print("\nGENERATED OUTPUT FILES")
print("-" * 90)

for key, file_path in output_files.items():
    print(f"{key}: {file_path}")

print(f"summary: {summary_path}")

print("\nGENERATED COMPARATIVE CHARTS")
print("-" * 90)

for figure_name in figure_files:
    print(OUTPUT_DIRS["figures"] / figure_name)

print("\n" + "=" * 90)

if failed_checks == 0:
    print("SHIPPING MODE EFFICIENCY ANALYSIS COMPLETED SUCCESSFULLY")
    print("ALL VALIDATION CHECKS PASSED")
else:
    print("SHIPPING MODE EFFICIENCY ANALYSIS COMPLETED WITH VALIDATION FAILURES")
    print("PLEASE REVIEW THE VALIDATION REPORT BEFORE PROCEEDING")

print("=" * 90)
```
