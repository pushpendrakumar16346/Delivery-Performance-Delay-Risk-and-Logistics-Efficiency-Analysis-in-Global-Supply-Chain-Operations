```python
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

warnings.filterwarnings("ignore")

print("=" * 95)
print("REGIONAL AND MARKET PERFORMANCE ANALYSIS")
print("=" * 95)


# LOAD AND INSPECT THE DELIVERY GAP DATASET

possible_roots = [
    Path.cwd(),
    *Path.cwd().parents,
    Path(r"C:\Users\hp\Analysis in Global Supply Chain Operations")
]

PROJECT_ROOT = None

for candidate in possible_roots:
    expected_file = candidate / "data" / "processed" / "delivery_gap_analysis.csv"
    if expected_file.exists():
        PROJECT_ROOT = candidate
        break

if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Could not locate data\\processed\\delivery_gap_analysis.csv. "
        "Check the project directory."
    )

INPUT_FILE = PROJECT_ROOT / "data" / "processed" / "delivery_gap_analysis.csv"

OUTPUT_DIRS = {
    "tables": PROJECT_ROOT / "outputs" / "tables",
    "reports": PROJECT_ROOT / "outputs" / "reports",
    "figures": PROJECT_ROOT / "outputs" / "figures"
}

for directory in OUTPUT_DIRS.values():
    directory.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(INPUT_FILE)

required_columns = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Delivery Status",
    "Late_delivery_risk",
    "Order Region",
    "Order Country",
    "Market"
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )

print(f"\nProject directory: {PROJECT_ROOT}")
print(f"Input file: {INPUT_FILE}")
print(f"Dataset rows: {len(df):,}")
print(f"Dataset columns: {len(df.columns):,}")


# STANDARDIZE GEOGRAPHIC CATEGORIES AND VALIDATE DELIVERY RECORDS

analysis_df = df.copy()

geographic_columns = [
    "Order Region",
    "Order Country",
    "Market"
]

for column in geographic_columns:
    analysis_df[column] = (
        analysis_df[column]
        .astype("string")
        .str.strip()
        .replace({
            "": pd.NA,
            "nan": pd.NA,
            "None": pd.NA
        })
    )

analysis_df["Days for shipping (real)"] = pd.to_numeric(
    analysis_df["Days for shipping (real)"],
    errors="coerce"
)

analysis_df["Days for shipment (scheduled)"] = pd.to_numeric(
    analysis_df["Days for shipment (scheduled)"],
    errors="coerce"
)

analysis_df["Late_delivery_risk"] = pd.to_numeric(
    analysis_df["Late_delivery_risk"],
    errors="coerce"
)

analysis_df["Delivery Status"] = (
    analysis_df["Delivery Status"]
    .astype("string")
    .str.strip()
)

analysis_df["Delay Gap"] = (
    analysis_df["Days for shipping (real)"]
    - analysis_df["Days for shipment (scheduled)"]
)

analysis_df["Is Cancelled"] = (
    analysis_df["Delivery Status"]
    .str.casefold()
    .eq("shipping canceled")
    .fillna(False)
)

analysis_df["Valid Shipping Duration"] = (
    analysis_df["Days for shipping (real)"].notna()
    & analysis_df["Days for shipment (scheduled)"].notna()
    & (analysis_df["Days for shipping (real)"] >= 0)
    & (analysis_df["Days for shipment (scheduled)"] >= 0)
)

analysis_df["Eligible Delivery"] = (
    ~analysis_df["Is Cancelled"]
    & analysis_df["Valid Shipping Duration"]
)

analysis_df["Delivery Classification"] = np.select(
    [
        analysis_df["Delay Gap"] > 0,
        analysis_df["Delay Gap"] == 0,
        analysis_df["Delay Gap"] < 0
    ],
    [
        "Delayed",
        "On-time",
        "Early"
    ],
    default="Invalid"
)

analysis_df.loc[
    ~analysis_df["Eligible Delivery"],
    "Delivery Classification"
] = "Ineligible"

eligible_df = analysis_df[
    analysis_df["Eligible Delivery"]
].copy()

if eligible_df.empty:
    raise ValueError(
        "No eligible delivery records are available for analysis."
    )

print("\nDATA VALIDATION SUMMARY")
print("-" * 95)
print(f"Total records: {len(analysis_df):,}")
print(f"Eligible delivery records: {len(eligible_df):,}")
print(
    f"Cancelled shipments: {int(analysis_df['Is Cancelled'].sum()):,}"
)
print(
    "Invalid or incomplete duration records:",
    int((~analysis_df["Valid Shipping Duration"]).sum())
)


# IDENTIFY AVAILABLE REGIONS, COUNTRIES AND MARKETS

regions = sorted(
    eligible_df["Order Region"].dropna().unique().tolist()
)

countries = sorted(
    eligible_df["Order Country"].dropna().unique().tolist()
)

markets = sorted(
    eligible_df["Market"].dropna().unique().tolist()
)

print("\nAVAILABLE GEOGRAPHIC CATEGORIES")
print("-" * 95)
print(f"Order Regions: {len(regions)}")
print(f"Order Countries: {len(countries)}")
print(f"Markets: {len(markets)}")

print("\nOrder Region categories:")
print(regions)

print("\nMarket categories:")
print(markets)


# CREATE REUSABLE GEOGRAPHIC PERFORMANCE CALCULATION

def calculate_geographic_performance(data, grouping_column):

    records = []

    for category, group in data.groupby(
        grouping_column,
        dropna=True,
        sort=True
    ):

        shipment_count = len(group)

        delayed_count = int(
            (group["Delivery Classification"] == "Delayed").sum()
        )

        on_time_count = int(
            (group["Delivery Classification"] == "On-time").sum()
        )

        early_count = int(
            (group["Delivery Classification"] == "Early").sum()
        )

        delayed_percentage = (
            delayed_count / shipment_count * 100
            if shipment_count else np.nan
        )

        on_time_percentage = (
            on_time_count / shipment_count * 100
            if shipment_count else np.nan
        )

        early_percentage = (
            early_count / shipment_count * 100
            if shipment_count else np.nan
        )

        risk_values = group["Late_delivery_risk"].dropna()
        risk_values = risk_values[risk_values.isin([0, 1])]

        risk_ratio = (
            risk_values.mean() * 100
            if len(risk_values) else np.nan
        )

        sla_compliant_count = int(
            (group["Delay Gap"] <= 0).sum()
        )

        sla_percentage = (
            sla_compliant_count / shipment_count * 100
            if shipment_count else np.nan
        )

        average_gap = group["Delay Gap"].mean()
        median_gap = group["Delay Gap"].median()

        delayed_records = group[
            group["Delivery Classification"] == "Delayed"
        ]

        average_delay = delayed_records["Delay Gap"].mean()

        maximum_delay = delayed_records["Delay Gap"].max()

        cancelled_count = int(
            (
                analysis_df[grouping_column].eq(category)
                & analysis_df["Is Cancelled"]
            ).sum()
        )

        records.append({
            grouping_column: category,
            "Eligible Shipments": shipment_count,
            "Cancelled Shipments": cancelled_count,
            "Delayed Shipments": delayed_count,
            "On-time Shipments": on_time_count,
            "Early Shipments": early_count,
            "Delayed Percentage": delayed_percentage,
            "On-time Percentage": on_time_percentage,
            "Early Delivery Percentage": early_percentage,
            "Average Delay Gap": average_gap,
            "Median Delay Gap": median_gap,
            "Average Delay Among Delayed Shipments": average_delay,
            "Maximum Delay Gap": maximum_delay,
            "Late Delivery Risk Ratio": risk_ratio,
            "Risk Records Available": int(len(risk_values)),
            "SLA Compliant Shipments": sla_compliant_count,
            "Regional SLA Compliance": sla_percentage
        })

    return pd.DataFrame(records)


# ANALYSE REGIONAL SHIPMENT VOLUMES AND DELIVERY PERFORMANCE

regional_analysis = calculate_geographic_performance(
    eligible_df,
    "Order Region"
)

regional_analysis = regional_analysis.sort_values(
    ["Delayed Percentage", "Average Delay Gap"],
    ascending=[False, False]
).reset_index(drop=True)

regional_analysis["Delay Intensity Rank"] = (
    regional_analysis["Delayed Percentage"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

print("\nREGIONAL DELIVERY PERFORMANCE")
print("-" * 95)
display(regional_analysis.round(3))


# RANK REGIONS BY DELAY INTENSITY AND IDENTIFY HIGH-RISK REGIONS

regional_threshold = regional_analysis[
    "Delayed Percentage"
].quantile(0.75)

regional_analysis["Regional Risk Category"] = np.where(
    regional_analysis["Delayed Percentage"] >= regional_threshold,
    "High Risk",
    "Lower Relative Risk"
)

high_risk_regions = regional_analysis[
    regional_analysis["Regional Risk Category"] == "High Risk"
].copy()

print("\nREGIONS RANKED BY DELAY INTENSITY")
print("-" * 95)

display(
    regional_analysis[
        [
            "Delay Intensity Rank",
            "Order Region",
            "Eligible Shipments",
            "Delayed Shipments",
            "Delayed Percentage",
            "Average Delay Gap",
            "Late Delivery Risk Ratio",
            "Regional SLA Compliance",
            "Regional Risk Category"
        ]
    ].round(3)
)

print("\nHIGH-RISK REGIONS")
print("-" * 95)

if high_risk_regions.empty:
    print("No high-risk regions identified.")
else:
    display(
        high_risk_regions[
            [
                "Order Region",
                "Eligible Shipments",
                "Delayed Percentage",
                "Average Delay Gap",
                "Late Delivery Risk Ratio",
                "Regional SLA Compliance"
            ]
        ].round(3)
    )


# ANALYSE COUNTRY-WISE SHIPMENT VOLUMES AND DELAY PERFORMANCE

country_analysis = calculate_geographic_performance(
    eligible_df,
    "Order Country"
)

country_analysis = country_analysis.sort_values(
    ["Delayed Percentage", "Average Delay Gap"],
    ascending=[False, False]
).reset_index(drop=True)

country_analysis["Country Delay Rank"] = (
    country_analysis["Delayed Percentage"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

country_threshold = country_analysis[
    "Delayed Percentage"
].quantile(0.75)

country_analysis["Country Risk Category"] = np.where(
    country_analysis["Delayed Percentage"] >= country_threshold,
    "High Risk",
    "Lower Relative Risk"
)

high_risk_countries = country_analysis[
    country_analysis["Country Risk Category"] == "High Risk"
].copy()

print("\nCOUNTRY-WISE DELIVERY PERFORMANCE")
print("-" * 95)

display(
    country_analysis[
        [
            "Country Delay Rank",
            "Order Country",
            "Eligible Shipments",
            "Delayed Shipments",
            "Delayed Percentage",
            "Average Delay Gap",
            "Late Delivery Risk Ratio",
            "Regional SLA Compliance",
            "Country Risk Category"
        ]
    ].round(3)
)

print("\nHIGH-RISK COUNTRIES")
print("-" * 95)

display(
    high_risk_countries[
        [
            "Order Country",
            "Eligible Shipments",
            "Delayed Shipments",
            "Delayed Percentage",
            "Average Delay Gap",
            "Late Delivery Risk Ratio"
        ]
    ].round(3)
)


# ANALYSE MARKET-WISE LOGISTICS PERFORMANCE

market_analysis = calculate_geographic_performance(
    eligible_df,
    "Market"
)

market_analysis = market_analysis.sort_values(
    ["Regional SLA Compliance", "Delayed Percentage"],
    ascending=[False, True]
).reset_index(drop=True)

market_analysis["Market Performance Rank"] = (
    market_analysis["Regional SLA Compliance"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

market_analysis["Market Delay Rank"] = (
    market_analysis["Delayed Percentage"]
    .rank(method="min", ascending=True)
    .astype("Int64")
)

print("\nMARKET-WISE LOGISTICS PERFORMANCE")
print("-" * 95)

display(
    market_analysis[
        [
            "Market Performance Rank",
            "Market Delay Rank",
            "Market",
            "Eligible Shipments",
            "Delayed Shipments",
            "Delayed Percentage",
            "On-time Percentage",
            "Average Delay Gap",
            "Late Delivery Risk Ratio",
            "Regional SLA Compliance"
        ]
    ].round(3)
)


# GENERATE MARKET LOGISTICS PERFORMANCE INDEX

gap_min = market_analysis["Average Delay Gap"].min()
gap_max = market_analysis["Average Delay Gap"].max()

if np.isclose(gap_min, gap_max):
    market_analysis["Delay Gap Efficiency Score"] = 100.0
else:
    market_analysis["Delay Gap Efficiency Score"] = (
        (gap_max - market_analysis["Average Delay Gap"])
        / (gap_max - gap_min)
    ) * 100

market_analysis["Risk Efficiency Score"] = (
    100 - market_analysis["Late Delivery Risk Ratio"]
)

market_analysis["Market Logistics Efficiency Index"] = (
    0.50 * market_analysis["Regional SLA Compliance"]
    + 0.30 * market_analysis["Risk Efficiency Score"]
    + 0.20 * market_analysis["Delay Gap Efficiency Score"]
)

market_analysis["Market Logistics Rank"] = (
    market_analysis["Market Logistics Efficiency Index"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

market_ranking = market_analysis[
    [
        "Market Logistics Rank",
        "Market",
        "Eligible Shipments",
        "Delayed Percentage",
        "On-time Percentage",
        "Average Delay Gap",
        "Late Delivery Risk Ratio",
        "Regional SLA Compliance",
        "Market Logistics Efficiency Index"
    ]
].sort_values(
    ["Market Logistics Rank", "Market"]
).reset_index(drop=True)

print("\nMARKET LOGISTICS EFFICIENCY RANKING")
print("-" * 95)
print(
    "Index formula: 50% SLA compliance + 30% inverse late delivery risk "
    "+ 20% relative average delay-gap score."
)
display(market_ranking.round(3))


# PREPARE REGIONAL GEOGRAPHIC MAPPING DATA

regional_mapping_data = regional_analysis[
    [
        "Order Region",
        "Eligible Shipments",
        "Delayed Shipments",
        "Delayed Percentage",
        "On-time Percentage",
        "Average Delay Gap",
        "Late Delivery Risk Ratio",
        "Regional SLA Compliance",
        "Regional Risk Category"
    ]
].copy()

regional_mapping_data["Geographic Level"] = "Order Region"

print("\nREGIONAL GEOGRAPHIC MAPPING DATA")
print("-" * 95)
display(regional_mapping_data.round(3))


# PREPARE COUNTRY-LEVEL GEOGRAPHIC MAPPING DATA

country_mapping_data = country_analysis[
    [
        "Order Country",
        "Eligible Shipments",
        "Delayed Shipments",
        "Delayed Percentage",
        "On-time Percentage",
        "Average Delay Gap",
        "Late Delivery Risk Ratio",
        "Regional SLA Compliance",
        "Country Risk Category"
    ]
].copy()

country_mapping_data["Geographic Level"] = "Order Country"

print("\nCOUNTRY-LEVEL GEOGRAPHIC MAPPING DATA")
print("-" * 95)
display(country_mapping_data.head(30).round(3))

print(
    "\nGeographic mapping limitation: the dataset contains Order Country "
    "and Order Region but does not provide verified destination coordinates. "
    "The mapping datasets are geographic performance summaries, not "
    "coordinate-based maps."
)


# PREPARE REGIONAL HEATMAP DATA

regional_heatmap_data = regional_analysis.set_index(
    "Order Region"
)[
    [
        "Delayed Percentage",
        "On-time Percentage",
        "Early Delivery Percentage",
        "Late Delivery Risk Ratio",
        "Regional SLA Compliance",
        "Average Delay Gap"
    ]
]

plt.figure(figsize=(13, 9))

sns.heatmap(
    regional_heatmap_data,
    annot=True,
    fmt=".2f",
    cmap="RdYlGn_r",
    linewidths=0.5
)

plt.title("Regional Delivery Risk and Logistics Performance Heatmap")
plt.xlabel("Performance Indicators")
plt.ylabel("Order Region")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "regional_delivery_performance_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# PREPARE COUNTRY-LEVEL HEATMAP DATA

# All countries remain available in the exported mapping table.
# The chart displays the 30 countries with the highest eligible shipment volumes.

top_countries = country_analysis.nlargest(
    min(30, len(country_analysis)),
    "Eligible Shipments"
).copy()

country_heatmap_data = top_countries.set_index(
    "Order Country"
)[
    [
        "Delayed Percentage",
        "On-time Percentage",
        "Late Delivery Risk Ratio",
        "Regional SLA Compliance",
        "Average Delay Gap"
    ]
]

plt.figure(figsize=(14, 12))

sns.heatmap(
    country_heatmap_data,
    annot=True,
    fmt=".1f",
    cmap="RdYlGn_r",
    linewidths=0.3
)

plt.title("Country-Level Delivery Performance Heatmap - Top 30 by Volume")
plt.xlabel("Performance Indicators")
plt.ylabel("Order Country")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "country_delivery_performance_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# VISUALIZE HIGH-RISK REGIONS

high_risk_region_plot = regional_analysis.sort_values(
    "Delayed Percentage",
    ascending=False
)

plt.figure(figsize=(14, 8))

sns.barplot(
    data=high_risk_region_plot,
    x="Delayed Percentage",
    y="Order Region"
)

plt.axvline(
    regional_threshold,
    linestyle="--",
    linewidth=1,
    label="High-risk threshold"
)

plt.title("Regional Delay Risk Comparison")
plt.xlabel("Delayed Shipments (%)")
plt.ylabel("Order Region")
plt.legend()
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "high_risk_regional_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# VISUALIZE HIGH-RISK COUNTRIES

high_risk_country_plot = country_analysis.nlargest(
    min(20, len(country_analysis)),
    "Delayed Percentage"
).sort_values(
    "Delayed Percentage",
    ascending=True
)

plt.figure(figsize=(13, 9))

sns.barplot(
    data=high_risk_country_plot,
    x="Delayed Percentage",
    y="Order Country"
)

plt.title("Highest Delay Percentage by Country - Top 20")
plt.xlabel("Delayed Shipments (%)")
plt.ylabel("Order Country")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "high_risk_country_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# CREATE MARKET-WISE COMPARISON CHARTS

market_plot = market_analysis.sort_values(
    "Market"
).copy()

plt.figure(figsize=(11, 6))

sns.barplot(
    data=market_plot,
    x="Market",
    y="Delayed Percentage"
)

plt.title("Market-Wise Delayed Shipment Percentage")
plt.xlabel("Market")
plt.ylabel("Delayed Shipments (%)")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "market_wise_delay_percentage.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


plt.figure(figsize=(11, 6))

sns.barplot(
    data=market_plot,
    x="Market",
    y="Late Delivery Risk Ratio"
)

plt.title("Market-Wise Late Delivery Risk Ratio")
plt.xlabel("Market")
plt.ylabel("Late Delivery Risk (%)")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "market_wise_late_delivery_risk.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


plt.figure(figsize=(11, 6))

sns.barplot(
    data=market_plot,
    x="Market",
    y="Regional SLA Compliance"
)

plt.title("Market-Wise SLA Compliance")
plt.xlabel("Market")
plt.ylabel("SLA Compliance (%)")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "market_wise_sla_compliance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


plt.figure(figsize=(11, 6))

sns.barplot(
    data=market_ranking,
    x="Market",
    y="Market Logistics Efficiency Index"
)

plt.title("Market Logistics Efficiency Index Comparison")
plt.xlabel("Market")
plt.ylabel("Efficiency Index (0-100)")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIRS["figures"] / "market_logistics_efficiency_index.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# IDENTIFY THE HIGHEST-RISK GEOGRAPHIC AREAS

highest_delay_region = regional_analysis.sort_values(
    "Delayed Percentage",
    ascending=False
).iloc[0]

highest_delay_country = country_analysis.sort_values(
    "Delayed Percentage",
    ascending=False
).iloc[0]

highest_risk_market = market_analysis.sort_values(
    "Late Delivery Risk Ratio",
    ascending=False
).iloc[0]

best_market = market_ranking.iloc[0]

print("\nGEOGRAPHIC PERFORMANCE FINDINGS")
print("-" * 95)

print(
    f"Highest-delay region: {highest_delay_region['Order Region']} "
    f"({highest_delay_region['Delayed Percentage']:.2f}%)"
)

print(
    f"Highest-delay country: {highest_delay_country['Order Country']} "
    f"({highest_delay_country['Delayed Percentage']:.2f}%)"
)

print(
    f"Highest-risk market: {highest_risk_market['Market']} "
    f"({highest_risk_market['Late Delivery Risk Ratio']:.2f}%)"
)

print(
    f"Best-performing market: {best_market['Market']} "
    f"(Efficiency Index: {best_market['Market Logistics Efficiency Index']:.2f})"
)


# GENERATE REGIONAL DIAGNOSTICS REPORT

regional_diagnostics_report = regional_analysis.copy()

regional_diagnostics_report["Operational Interpretation"] = np.where(
    regional_diagnostics_report["Regional Risk Category"] == "High Risk",
    "Requires regional delivery performance review",
    "Lower relative delay risk compared with other regions"
)

country_diagnostics_report = country_analysis.copy()

country_diagnostics_report["Operational Interpretation"] = np.where(
    country_diagnostics_report["Country Risk Category"] == "High Risk",
    "Requires country-level delivery performance review",
    "Lower relative delay risk compared with other countries"
)

market_diagnostics_report = market_analysis.copy()

market_diagnostics_report["Operational Interpretation"] = np.where(
    market_diagnostics_report["Market Logistics Efficiency Index"]
    >= market_diagnostics_report["Market Logistics Efficiency Index"].median(),
    "At or above the median relative market efficiency",
    "Below the median relative market efficiency"
)

print("\nREGIONAL DIAGNOSTICS REPORT")
print("-" * 95)

display(
    regional_diagnostics_report[
        [
            "Order Region",
            "Eligible Shipments",
            "Delayed Percentage",
            "Average Delay Gap",
            "Late Delivery Risk Ratio",
            "Regional SLA Compliance",
            "Regional Risk Category",
            "Operational Interpretation"
        ]
    ].round(3)
)


# SAVE REGIONAL, COUNTRY AND MARKET ANALYSIS OUTPUTS

output_files = {
    "regional_analysis":
        OUTPUT_DIRS["tables"] / "regional_delivery_performance.csv",

    "regional_ranking":
        OUTPUT_DIRS["tables"] / "regional_delay_intensity_ranking.csv",

    "high_risk_regions":
        OUTPUT_DIRS["tables"] / "high_risk_regions.csv",

    "country_analysis":
        OUTPUT_DIRS["tables"] / "country_delivery_performance.csv",

    "country_ranking":
        OUTPUT_DIRS["tables"] / "country_delay_ranking.csv",

    "high_risk_countries":
        OUTPUT_DIRS["tables"] / "high_risk_countries.csv",

    "market_analysis":
        OUTPUT_DIRS["tables"] / "market_delivery_performance.csv",

    "market_ranking":
        OUTPUT_DIRS["tables"] / "market_logistics_efficiency_ranking.csv",

    "regional_mapping":
        OUTPUT_DIRS["tables"] / "regional_geographic_mapping_data.csv",

    "country_mapping":
        OUTPUT_DIRS["tables"] / "country_geographic_mapping_data.csv",

    "regional_heatmap_data":
        OUTPUT_DIRS["tables"] / "regional_heatmap_data.csv",

    "country_heatmap_data":
        OUTPUT_DIRS["tables"] / "country_heatmap_data.csv",

    "regional_report":
        OUTPUT_DIRS["reports"] / "regional_diagnostics_report.csv",

    "country_report":
        OUTPUT_DIRS["reports"] / "country_diagnostics_report.csv",

    "market_report":
        OUTPUT_DIRS["reports"] / "market_diagnostics_report.csv",

    "validation":
        OUTPUT_DIRS["reports"] / "regional_market_analysis_validation.csv"
}

regional_analysis.to_csv(
    output_files["regional_analysis"], index=False
)

regional_analysis.to_csv(
    output_files["regional_ranking"], index=False
)

high_risk_regions.to_csv(
    output_files["high_risk_regions"], index=False
)

country_analysis.to_csv(
    output_files["country_analysis"], index=False
)

country_analysis.to_csv(
    output_files["country_ranking"], index=False
)

high_risk_countries.to_csv(
    output_files["high_risk_countries"], index=False
)

market_analysis.to_csv(
    output_files["market_analysis"], index=False
)

market_ranking.to_csv(
    output_files["market_ranking"], index=False
)

regional_mapping_data.to_csv(
    output_files["regional_mapping"], index=False
)

country_mapping_data.to_csv(
    output_files["country_mapping"], index=False
)

regional_heatmap_data.to_csv(
    output_files["regional_heatmap_data"]
)

country_heatmap_data.to_csv(
    output_files["country_heatmap_data"]
)

regional_diagnostics_report.to_csv(
    output_files["regional_report"], index=False
)

country_diagnostics_report.to_csv(
    output_files["country_report"], index=False
)

market_diagnostics_report.to_csv(
    output_files["market_report"], index=False
)


# VALIDATE REGIONAL AND MARKET ANALYSIS

validation_results = []

def add_validation(check_name, condition, details):
    validation_results.append({
        "Validation Check": check_name,
        "Status": "PASS" if bool(condition) else "FAIL",
        "Details": str(details)
    })


add_validation(
    "Input dataset loaded",
    len(analysis_df) == len(df) and len(df) > 0,
    f"Loaded {len(df):,} records."
)

add_validation(
    "Eligible delivery records identified",
    len(eligible_df) > 0
    and not eligible_df["Is Cancelled"].any(),
    f"Eligible records: {len(eligible_df):,}."
)

add_validation(
    "Regional categories identified",
    len(regions) > 0
    and regional_analysis["Order Region"].nunique() == len(regions),
    f"Regions identified: {len(regions)}."
)

add_validation(
    "Country categories identified",
    len(countries) > 0
    and country_analysis["Order Country"].nunique() == len(countries),
    f"Countries identified: {len(countries)}."
)

add_validation(
    "Market categories identified",
    len(markets) > 0
    and market_analysis["Market"].nunique() == len(markets),
    f"Markets identified: {len(markets)}."
)

add_validation(
    "Regional shipment counts reconcile",
    int(regional_analysis["Eligible Shipments"].sum())
    == int(eligible_df["Order Region"].notna().sum()),
    "Regional totals reconcile with eligible records having a region."
)

add_validation(
    "Country shipment counts reconcile",
    int(country_analysis["Eligible Shipments"].sum())
    == int(eligible_df["Order Country"].notna().sum()),
    "Country totals reconcile with eligible records having a country."
)

add_validation(
    "Market shipment counts reconcile",
    int(market_analysis["Eligible Shipments"].sum())
    == int(eligible_df["Market"].notna().sum()),
    "Market totals reconcile with eligible records having a market."
)

add_validation(
    "Regional delay classifications reconcile",
    int(regional_analysis["Delayed Shipments"].sum())
    + int(regional_analysis["On-time Shipments"].sum())
    + int(regional_analysis["Early Shipments"].sum())
    == int(eligible_df["Order Region"].notna().sum()),
    "Regional delayed, on-time and early counts reconcile."
)

add_validation(
    "Country delay classifications reconcile",
    int(country_analysis["Delayed Shipments"].sum())
    + int(country_analysis["On-time Shipments"].sum())
    + int(country_analysis["Early Shipments"].sum())
    == int(eligible_df["Order Country"].notna().sum()),
    "Country delayed, on-time and early counts reconcile."
)

add_validation(
    "Market delay classifications reconcile",
    int(market_analysis["Delayed Shipments"].sum())
    + int(market_analysis["On-time Shipments"].sum())
    + int(market_analysis["Early Shipments"].sum())
    == int(eligible_df["Market"].notna().sum()),
    "Market delayed, on-time and early counts reconcile."
)


def validate_geographic_percentages(result_df, category_column):

    checks = []

    for _, row in result_df.iterrows():

        denominator = row["Eligible Shipments"]

        if denominator <= 0:
            continue

        checks.append(
            np.isclose(
                row["Delayed Percentage"],
                row["Delayed Shipments"] / denominator * 100
            )
        )

        checks.append(
            np.isclose(
                row["On-time Percentage"],
                row["On-time Shipments"] / denominator * 100
            )
        )

        checks.append(
            np.isclose(
                row["Early Delivery Percentage"],
                row["Early Shipments"] / denominator * 100
            )
        )

        checks.append(
            np.isclose(
                row["Regional SLA Compliance"],
                row["SLA Compliant Shipments"] / denominator * 100
            )
        )

    return all(checks) if checks else False


add_validation(
    "Regional percentages use consistent eligible denominators",
    validate_geographic_percentages(
        regional_analysis, "Order Region"
    ),
    "Regional percentages use eligible shipments within each region."
)

add_validation(
    "Country percentages use consistent eligible denominators",
    validate_geographic_percentages(
        country_analysis, "Order Country"
    ),
    "Country percentages use eligible shipments within each country."
)

add_validation(
    "Market percentages use consistent eligible denominators",
    validate_geographic_percentages(
        market_analysis, "Market"
    ),
    "Market percentages use eligible shipments within each market."
)


def validate_risk_ratios(result_df, category_column):

    for _, row in result_df.iterrows():

        risk_values = eligible_df.loc[
            eligible_df[category_column] == row[category_column],
            "Late_delivery_risk"
        ]

        risk_values = risk_values[risk_values.isin([0, 1])]

        expected = (
            risk_values.mean() * 100
            if len(risk_values) else np.nan
        )

        actual = row["Late Delivery Risk Ratio"]

        if not np.isclose(actual, expected, equal_nan=True):
            return False

    return True


add_validation(
    "Regional risk ratios reconcile",
    validate_risk_ratios(regional_analysis, "Order Region"),
    "Regional risk ratios use valid binary risk records."
)

add_validation(
    "Country risk ratios reconcile",
    validate_risk_ratios(country_analysis, "Order Country"),
    "Country risk ratios use valid binary risk records."
)

add_validation(
    "Market risk ratios reconcile",
    validate_risk_ratios(market_analysis, "Market"),
    "Market risk ratios use valid binary risk records."
)

add_validation(
    "High-risk regions identified",
    len(high_risk_regions) > 0
    and set(high_risk_regions["Order Region"]).issubset(set(regions)),
    "High-risk regions are selected from identified regions."
)

add_validation(
    "High-risk countries identified",
    len(high_risk_countries) > 0
    and set(high_risk_countries["Order Country"]).issubset(set(countries)),
    "High-risk countries are selected from identified countries."
)

add_validation(
    "Regional ranking complete",
    regional_analysis["Delay Intensity Rank"].notna().all(),
    "All identified regions have a delay intensity rank."
)

add_validation(
    "Country ranking complete",
    country_analysis["Country Delay Rank"].notna().all(),
    "All identified countries have a delay rank."
)

add_validation(
    "Market ranking complete",
    market_ranking["Market Logistics Rank"].notna().all(),
    "All identified markets have an efficiency rank."
)

add_validation(
    "Market efficiency index within valid range",
    market_analysis[
        "Market Logistics Efficiency Index"
    ].between(0, 100).all(),
    "Market efficiency scores are within 0 and 100."
)

add_validation(
    "Cancelled shipments excluded from delivery metrics",
    not eligible_df["Is Cancelled"].any(),
    "Cancelled shipments are excluded from all eligible denominators."
)

add_validation(
    "Geographic mapping data prepared",
    (
        len(regional_mapping_data) == len(regional_analysis)
        and len(country_mapping_data) == len(country_analysis)
    ),
    "Regional and country mapping summaries are prepared."
)

add_validation(
    "Geographic coordinate limitation documented",
    (
        "Geographic Level" in regional_mapping_data.columns
        and "Geographic Level" in country_mapping_data.columns
    ),
    "Mapping data contains geographic categories and metrics, not invented coordinates."
)

# Verify saved analysis files.

for name, file_path in output_files.items():

    if name == "validation":
        continue

    add_validation(
        f"Output file exists: {file_path.name}",
        file_path.exists() and file_path.stat().st_size > 0,
        f"Verified: {file_path}"
    )

figure_files = [
    "regional_delivery_performance_heatmap.png",
    "country_delivery_performance_heatmap.png",
    "high_risk_regional_comparison.png",
    "high_risk_country_comparison.png",
    "market_wise_delay_percentage.png",
    "market_wise_late_delivery_risk.png",
    "market_wise_sla_compliance.png",
    "market_logistics_efficiency_index.png"
]

for figure_name in figure_files:

    figure_path = OUTPUT_DIRS["figures"] / figure_name

    add_validation(
        f"Chart exists: {figure_name}",
        figure_path.exists() and figure_path.stat().st_size > 0,
        f"Verified: {figure_path}"
    )

# Save the validation report before verifying its existence.

validation_report = pd.DataFrame(validation_results)

validation_report.to_csv(
    output_files["validation"],
    index=False
)

add_validation(
    "Output file exists: regional_market_analysis_validation.csv",
    output_files["validation"].exists()
    and output_files["validation"].stat().st_size > 0,
    f"Verified: {output_files['validation']}"
)

# Save the final report including its own file verification.

validation_report = pd.DataFrame(validation_results)

validation_report.to_csv(
    output_files["validation"],
    index=False
)

passed_checks = int(
    (validation_report["Status"] == "PASS").sum()
)

failed_checks = int(
    (validation_report["Status"] == "FAIL").sum()
)


# SAVE REGIONAL AND MARKET ANALYSIS SUMMARY

summary = {
    "project": "Analysis in Global Supply Chain Operations",
    "analysis": "Regional and Market Performance Analysis",
    "source_file": str(INPUT_FILE),
    "total_records": int(len(analysis_df)),
    "eligible_records": int(len(eligible_df)),
    "cancelled_records": int(analysis_df["Is Cancelled"].sum()),
    "region_count": int(len(regions)),
    "country_count": int(len(countries)),
    "market_count": int(len(markets)),
    "highest_delay_region": str(
        highest_delay_region["Order Region"]
    ),
    "highest_delay_region_percentage": float(
        highest_delay_region["Delayed Percentage"]
    ),
    "highest_delay_country": str(
        highest_delay_country["Order Country"]
    ),
    "highest_delay_country_percentage": float(
        highest_delay_country["Delayed Percentage"]
    ),
    "highest_risk_market": str(
        highest_risk_market["Market"]
    ),
    "best_performing_market": str(
        best_market["Market"]
    ),
    "regional_high_risk_threshold": float(regional_threshold),
    "country_high_risk_threshold": float(country_threshold),
    "geographic_coordinates_available": False,
    "geographic_mapping_note": (
        "Geographic performance summaries are available. "
        "Verified destination coordinates are not present in the dataset."
    ),
    "validation_passed": passed_checks,
    "validation_failed": failed_checks,
    "validation_status": (
        "PASS" if failed_checks == 0 else "FAIL"
    ),
    "output_files": {
        name: str(path)
        for name, path in output_files.items()
    },
    "figure_files": [
        str(OUTPUT_DIRS["figures"] / name)
        for name in figure_files
    ]
}

summary_path = (
    OUTPUT_DIRS["reports"] /
    "regional_market_analysis_summary.json"
)

with open(summary_path, "w", encoding="utf-8") as file:
    json.dump(
        summary,
        file,
        indent=4,
        ensure_ascii=False
    )


# DISPLAY FINAL REGIONAL AND MARKET RESULTS

print("\nFINAL REGIONAL DELAY RANKING")
print("=" * 95)

display(
    regional_analysis[
        [
            "Delay Intensity Rank",
            "Order Region",
            "Eligible Shipments",
            "Delayed Percentage",
            "Average Delay Gap",
            "Late Delivery Risk Ratio",
            "Regional SLA Compliance",
            "Regional Risk Category"
        ]
    ].round(3)
)

print("\nFINAL MARKET LOGISTICS RANKING")
print("=" * 95)

display(market_ranking.round(3))

print("\nGENERATED OUTPUT FILES")
print("-" * 95)

for name, file_path in output_files.items():
    print(f"{name}: {file_path}")

print(f"summary: {summary_path}")

print("\nGENERATED VISUALIZATIONS")
print("-" * 95)

for figure_name in figure_files:
    print(OUTPUT_DIRS["figures"] / figure_name)

print("\nVALIDATION SUMMARY")
print("-" * 95)
display(validation_report)

print(f"\nValidation checks passed: {passed_checks}")
print(f"Validation checks failed: {failed_checks}")

print("\n" + "=" * 95)

if failed_checks == 0:
    print("REGIONAL AND MARKET PERFORMANCE ANALYSIS COMPLETED SUCCESSFULLY")
    print("ALL VALIDATION CHECKS PASSED")
else:
    print("REGIONAL AND MARKET PERFORMANCE ANALYSIS COMPLETED WITH VALIDATION FAILURES")
    print("PLEASE REVIEW THE VALIDATION REPORT")

print("=" * 95)
```
